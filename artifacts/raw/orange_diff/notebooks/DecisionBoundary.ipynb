#### Decision Boundary per Digit

##### Intersection of KDEs

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

# Function to calculate KDEs and find intersection points, while plotting the distributions and threshold line
def find_kde_intersection(false_class_values, true_class_values):
    # Generate the KDE for both distributions
    kde_false = gaussian_kde(false_class_values)
    kde_true = gaussian_kde(true_class_values)
    
    # Create a range of values to evaluate the KDEs over (from min to max of both distributions)
    x_values = np.linspace(min(false_class_values.min(), true_class_values.min()), 
                           max(false_class_values.max(), true_class_values.max()), 1000)
    
    # Evaluate the KDEs
    kde_false_values = kde_false(x_values)
    kde_true_values = kde_true(x_values)
    
    # Find the intersection point (where the difference between the two KDEs is zero)
    diff = kde_false_values - kde_true_values
    indices = np.where(np.diff(np.sign(diff)))[0]
    
    # If intersection points exist, return the x-value of the intersection
    if len(indices) > 0:
        intersection_point = x_values[indices[0]]
        return intersection_point, x_values, kde_false_values, kde_true_values
    else:
        return None, x_values, kde_false_values, kde_true_values

# Set up a 2x5 grid for plotting
fig, axes = plt.subplots(2, 5, figsize=(20, 10))

# Flatten axes array for easier iteration
axes = axes.flatten()

# Find intersection-based threshold for all digits (0 to 9) and plot the KDEs with threshold line in subplots
best_thresholds = {}
for digit in range(10):
    false_class_values = filtered_false_class_dfs[digit].values
    true_class_values = filtered_true_class_dfs[digit].values
    
    # Calculate KDE intersection threshold and get the KDE values for plotting
    best_threshold, x_values, kde_false_values, kde_true_values = find_kde_intersection(false_class_values, true_class_values)
    best_thresholds[digit] = best_threshold
    
    # Plot in the corresponding subplot
    ax = axes[digit]
    
    # Plot KDEs for the false and true class with fill
    ax.fill_between(x_values, kde_false_values, color='red', alpha=0.5, label='False Class')
    ax.fill_between(x_values, kde_true_values, color='green', alpha=0.5, label='True Class')
    
    # Plot the intersection threshold as a vertical line
    if best_threshold is not None:
        ax.axvline(x=best_threshold, color='blue', linestyle='--', label=f'Intersection = {best_threshold:.2f}')
    
    # Add title and labels
    ax.set_title(f'Digit {digit} - KDE Intersection')
    ax.set_xlabel('Value')
    ax.set_ylabel('Density')
    ax.legend()

# Adjust layout so plots don't overlap
plt.tight_layout()
plt.show()
```

##### Midpoint of Means

```python
def midpoint_of_means(false_class_values, true_class_values):
    mean_false = np.mean(false_class_values)
    mean_true = np.mean(true_class_values)
    midpoint = (mean_false + mean_true) / 2
    return midpoint

for digit in range(10):
    false_class_values = filtered_false_class_dfs[digit].values
    true_class_values = filtered_true_class_dfs[digit].values
    best_threshold = midpoint_of_means(false_class_values, true_class_values)
    best_thresholds[digit] = best_threshold
    print(f"Best threshold for digit {digit} (intersection): {best_threshold}")
```

##### Midpoint of Medians

```python
def midpoint_of_medians(false_class_values, true_class_values):
    median_false = np.median(false_class_values)
    median_true = np.median(true_class_values)
    midpoint = (median_false + median_true) / 2
    return midpoint

for digit in range(10):
    false_class_values = filtered_false_class_dfs[digit].values
    true_class_values = filtered_true_class_dfs[digit].values
    best_threshold = midpoint_of_medians(false_class_values, true_class_values)
    best_thresholds[digit] = best_threshold
    print(f"Best threshold for digit {digit} (intersection): {best_threshold}")
```

##### Otsu's Method

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from skimage.filters import threshold_otsu

# Function to calculate Otsu threshold and plot the distributions with the threshold line
def otsu_threshold(false_class_values, true_class_values):
    # Combine the values
    combined_values = np.concatenate([false_class_values, true_class_values])
    
    # Calculate the Otsu threshold
    threshold = threshold_otsu(combined_values)
    
    return threshold

# Set up a 2x5 grid for plotting
fig, axes = plt.subplots(2, 5, figsize=(20, 10))

# Flatten axes array for easier iteration
axes = axes.flatten()

# Find Otsu threshold for all digits (0 to 9) and plot the KDEs with threshold line in subplots
best_thresholds = {}
for digit in range(10):
    false_class_values = filtered_false_class_dfs[digit].values
    true_class_values = filtered_true_class_dfs[digit].values
    
    # Calculate Otsu threshold
    best_threshold = otsu_threshold(false_class_values, true_class_values)
    best_thresholds[digit] = best_threshold
    
    # Plot in the corresponding subplot
    ax = axes[digit]
    sns.kdeplot(false_class_values, fill=True, color='red', label='False Class', ax=ax)
    sns.kdeplot(true_class_values, fill=True, color='green', label='True Class', ax=ax)
    
    # Plot the Otsu threshold as a vertical line
    ax.axvline(x=best_threshold, color='blue', linestyle='--', label=f'Otsu Threshold = {best_threshold:.2f}')
    
    # Add title and labels
    ax.set_title(f'Digit {digit} - Otsu Threshold')
    ax.set_xlabel('Value')
    ax.set_ylabel('Density')
    ax.legend()

# Adjust layout so plots don't overlap
plt.tight_layout()
plt.show()
```

##### Minimizing Total Error

```python
def find_best_threshold(false_class_values, true_class_values):
    combined_values = np.sort(np.concatenate([false_class_values, true_class_values]))
    best_threshold = None
    best_error = float('inf')
    
    for i in range(len(combined_values) - 1):
        threshold = (combined_values[i] + combined_values[i + 1]) / 2
        false_negatives = np.sum(true_class_values <= threshold)  # True class below threshold
        false_positives = np.sum(false_class_values > threshold)  # False class above threshold
        total_error = false_negatives + false_positives
        
        if total_error < best_error:
            best_threshold = threshold
            best_error = total_error
    
    return best_threshold

for digit in range(10):
    false_class_values = filtered_false_class_dfs[digit].values
    true_class_values = filtered_true_class_dfs[digit].values
    best_threshold = find_best_threshold(false_class_values, true_class_values)
    best_thresholds[digit] = best_threshold
    print(f"Best threshold for digit {digit} (intersection): {best_threshold}")
```
